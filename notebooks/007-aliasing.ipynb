## Setup

This notebook depicts the aliasing and anti aliasing using sine wave.

```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
samples = np.linspace(-300, 300, num=100)
sine_samples = np.sin(samples)
```

## Sine Signal

```python
plt.plot(sine_samples, label='sine wave', lw=3)
plt.scatter(range(len(samples)), sine_samples, label='points', c='r')
plt.xlabel('Sample Number'); plt.ylabel('sine values'); plt.title('Correct Sine Signal'); plt.legend()
```

## Aliasing

```python
aliased_sine_samples = np.sin(np.linspace(-300,300,10))
```

```python
plt.plot(sine_samples, label='sine wave', lw=3)
plt.plot(aliased_sine_samples, label='alisaed sine wave', lw=3, c='k')
plt.scatter(range(len(aliased_sine_samples)), aliased_sine_samples, label='aliased points', c='g')
plt.xlabel('Sample Number'); plt.ylabel('sine values'); plt.title('Correct Sine Signal'); plt.legend()
```
